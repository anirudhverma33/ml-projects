```python
#execution performance
size = 10_000_000

#python list
py_list = list(range(size))
start = time.time()
sq_list = [x**2 for x in py_list]
end = time.time()
print(f"python list time = {end -start} seconds")


#numpy arrays
np_arr = np.array(py_list)
start = time.time()
#vectorization 
sq_array = np_arr**2
end = time.time()

print(f"numpy array time = {end -start} seconds")
```

```python
#indexing 
# arr = np.array ([1,2,3,4,5])

arr = np.array([[1,2,3],[4,5,6]])
print(arr[0][1])
print(arr[1][1])
```

```python
# Fancy Indexing
arr = np.array ([1,2,3,4,5])

idx = [0,1,3]
print(arr[idx])
```

```python
# Boolean Indexing 

arr = np.array ([1,2,3,4,5])
print(arr[arr%2 ==0])
```

```python
#Slicing 
arr = np.array ([1,2,3,4,5])
print(arr[1:4]) #[start,end,step]
print(arr[1:4:2])
```

```python
# There is a big difference how slicing is done memory wise

# Copy vs View

# nums  =[1,2,3,4,5]
# sub_list = nums[1:3]
# print(sub_list)
# sub_list[0] = 200

# print(sub_list)
# print(nums)


nums  = np.array ([1,2,3,4,5])
sub_list = nums[1:3].copy()
print(sub_list)
sub_list[0] = 200

print(sub_list)
print(nums)
#No change in copy 
```

```python
#normalise

arr = np.array([[1,2],[3,4]])

mean = np.mean(arr)
std_dev = np.std(arr)

normalised_arr = (arr-mean )/std_dev
print(normalised_arr)

print(mean)
print(std_dev)
```

```python
#Mathematical Functions

#Aggregation

arr = np.array([1,2,3,4,5])

print(np.sum(arr))
print(np.prod(arr))
print(np.min(arr))
print(np.max(arr))

print(np.argmin(arr))
print(np.argmax(arr))

print(np.mean(arr))
print(np.std(arr))
print(np.median(arr))
```

```python
#Power Functions
arr = np.array([1,2,3,4,5])

print(np.square(arr))
print(np.sqrt(arr))
print(np.power(arr,3))
```

```python
#log
arr = np.array([1,2,3,4,5])

print(np.log(arr))
print(np.log10(arr))
print(np.exp(arr))
```

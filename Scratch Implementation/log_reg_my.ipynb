```python
import numpy as np
```

```python
# Logistic Regression 

class LogisticRegression:
    
    def __init__(self, lr = 0.02, n_iter= 1000):
        self.lr = lr
        self.n_iter = n_iter
        self.bias = None
        self.weights = None

    def _sigmoid(self,z):
        return (1/(1 + np.exp(-z)))

    def fit(self, X, y):
        m,n = X.shape
        self.bias = 0
        self.weights = np.zeros(n)


        for _ in range(self.n_iter):
            y_pr = self.bias + X@self.weights
            y_pred = self._sigmoid(y_pr)
            db = 1/m * (y_pred - y)
            dw = 1/m * X.T @ (y_pred - y)

            self.bias -=self.lr*db
            self.weights -= self.lr*dw

    def prob(self,X):
        z = self.bias + X@self.weights
        return self._sigmoid(z)

    def predict(self,X, th = 0.5):
        probs = self.prob(X)
        prob_bool = probs>= th
        return prob_bool.astype(int)
```

```python
X = np.array([[1, 2], [2, 3], [3, 4], [4, 5]])
y = np.array([0, 0, 1, 1])

model = LogisticRegression()
model.fit(X, y)

y_pred = model.predict(X)
print(y_pred)

print(model.prob(X))
```

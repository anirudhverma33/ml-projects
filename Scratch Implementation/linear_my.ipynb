```python
#Linear Regression 

class LinearRegression1:
    def __init__(self, learning_rate=0.01, n_iters=1000):
        self.learning_rate = learning_rate
        self.n_iters = n_iters
        self.weights = None
        self.bias = None

    def fit(self, X, y):
        m,n = X.shape # (samples, features)

        # initialize parameters 
        self.weights = np.zeros(n)
        self.bias = 0

        # gradient descent
        for _ in range(self.n_iters):
            y_predicted = np.dot(X, self.weights) + self.bias

            dw = (1/m) * np.dot(X.T, (y_predicted - y))
            db = (1/m) * np.sum(y_predicted - y)

            self.weights -= self.learning_rate * dw
            self.bias -= self.learning_rate * db

    def predict(self, X):
        y_approximated = np.dot(X, self.weights) + self.bias
        return y_approximated
    

```

```python
class LinearRegression:
    def __init__(self, learning_rate=0.01, n_iters=1000):
        self.learning_rate = learning_rate
        self.n_iters = n_iters
        self.weights = None
        self.bias = None

    def fit(self, X, y):
       m,n = X.shape # (samples, features)

       #Step 1 is assigning weight and bias 
       self.weights = np.zeros(n)
       self.bias = 0

       for _ in range(self.n_iters): 
        #Step 2 is gradient descent
        y_predicted = self.bias + X@self.weights
        
        dw = 1/m * X.T@(y_predicted - y)
        db = 1/m * np.sum(y_predicted - y)

        #Step 3 is updating weights and bias
        self.weights -= self.learning_rate * dw
        self.bias -= self.learning_rate * db

    def predict(self, X):
        y_approximated = self.bias + X@self.weights
        return y_approximated
    
```

```python
X = np.array([[1, 2], [2, 4], [3, 6], [4, 8], [5, 10]])
y = np.array([2, 4, 6, 8, 10])

model = LinearRegression()
model.fit(X, y)

y_pred = model.predict(X)
print(y_pred)

print(model.bias)
print(model.weights)
```

```python

#Linear Regression with OLS

class LinearRegressionOLS:

    def __init__(self):
        self.weights = None
        self.bias = None

    def fit(self, X, y):
        m, n = X.shape

        # Add a column of ones to X for the bias term
        X_b = np.hstack([np.ones((m, 1)), X])

        # Calculate weights using the normal equation
        theta = np.linalg.inv(X_b.T @ X_b) @ X_b.T @ y

        self.bias = theta[0]
        self.weights = theta[1:]


    def predict(self, X):
        y_pred = self.bias + X @ self.weights
        return y_pred
    
```

```python
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

model = LinearRegressionOLS()
model.fit(X, y)

y_pred = model.predict(X)
print(y_pred)

print(model.bias)
print(model.weights)
```
